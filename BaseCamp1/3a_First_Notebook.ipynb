A Jupyter notebook runs on a single **kernel** — one live Python process. Every cell you run executes inside that same process, so variables defined in one cell stay in memory and are available to every cell that runs after it.

Run the cells **in order, top to bottom**

````python
import sys, os

print("Python executable:", sys.executable)
print("Python version:", sys.version.split()[0])
print("Process ID:", os.getpid())
````

### Define a variable

Run the cell below. It creates two variables, `message` and `counter`.

````python
message = "Hello from cell n"
counter = 0

print("Defined:", message, "| counter =", counter)
````

### Reuse it in a later cell

Notice this next cell never redefines `message` or `counter` — it just uses them. They're still sitting in the kernel's memory from the previous cell.

````python
counter += 1
print("This is cell n+1. Message exist : ", message)
print("counter is now:", counter)
````

### Persistence keeps accumulating

Run the cell below a few times in a row (Shift+Enter repeatedly) and watch `counter` keep climbing — the state carries forward every time.

````python
counter += 1
history = [message] * counter

print("counter is now:", counter)
print("history:", history)
````

## What breaks the persistence

The variables live in the **kernel**, not in the notebook file itself. If you restart the kernel (Kernel → Restart) and then run *only* the cell below without re-running the earlier ones, it will fail:

```
NameError: name 'message' is not defined
```

Try it — restart the kernel, then run just this one cell.

````python
print(message)
print("counter is now:", counter)
````
