```python
def divide(*args):
    if len(args) != 2:
        raise ValueError("Function requires exactly two arguments.")
    a, b = args
    if b == 0:
        raise ValueError("Cannot divide by zero.")
    return a / b
```

```python
divide(1,0)
```

```python
def divide(a, b):
    try:
        return_value = a / b
        return return_value
    except Exception as exception:
        return "An error occurred: " + str(exception)
```

```python
divide(1,0)
```
